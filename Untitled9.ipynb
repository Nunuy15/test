```r
# Membaca dataset
data <- read.csv("/content/Social_Network_Ads.csv")

# Mean, Median, dan Modus
mean_age <- mean(data$Age)
median_age <- median(data$Age)
mode_age <- as.numeric(names(sort(table(data$Age), decreasing = TRUE)[1]))

mean_salary <- mean(data$EstimatedSalary)
median_salary <- median(data$EstimatedSalary)
mode_salary <- as.numeric(names(sort(table(data$EstimatedSalary), decreasing = TRUE)[1]))

cat("a. Statistik Tendensi Sentral\n")
cat("Age - Mean:", mean_age, ", Median:", median_age, ", Mode:", mode_age, "\n")
cat("EstimatedSalary - Mean:", mean_salary, ", Median:", median_salary, ", Mode:", mode_salary, "\n\n")

# Jangkauan dan Kuartil
range_age <- range(data$Age)
range_salary <- range(data$EstimatedSalary)

quartiles_age <- quantile(data$Age)
quartiles_salary <- quantile(data$EstimatedSalary)

cat("b. Jangkauan dan Kuartil\n")
cat("Age - Range:", diff(range_age), ", Q1:", quartiles_age[2], ", Q2 (Median):", quartiles_age[3], ", Q3:", quartiles_age[4], "\n")
cat("EstimatedSalary - Range:", diff(range_salary), ", Q1:", quartiles_salary[2], ", Q2 (Median):", quartiles_salary[3], ", Q3:", quartiles_salary[4], "\n\n")

# Varians dan Standar Deviasi
var_age <- var(data$Age)
sd_age <- sd(data$Age)

var_salary <- var(data$EstimatedSalary)
sd_salary <- sd(data$EstimatedSalary)

cat("c. Varians dan Standar Deviasi\n")
cat("Age - Variance:", var_age, ", Standard Deviation:", sd_age, "\n")
cat("EstimatedSalary - Variance:", var_salary, ", Standard Deviation:", sd_salary, "\n")
```
